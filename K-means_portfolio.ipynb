# FIFA 22 Player Segmentation (Portfolio Notebook)

This notebook builds a **production-style K-Means workflow** with:
- Data validation checks
- Outlier analysis
- Robust preprocessing
- K selection using multiple metrics
- Cluster interpretation and visualization

Dataset used: `data/players_22.csv`


```python
# Segment 1: Imports and global settings
# -------------------------------------
# We import all libraries required for data processing, validation,
# clustering, metrics, and visualization.

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.preprocessing import RobustScaler
from sklearn.decomposition import PCA

# Make plots easier to read.
sns.set(style="whitegrid", context="notebook")
pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 140)

RANDOM_STATE = 42
```

```python
# Segment 2: Load data and run basic structure checks
# ---------------------------------------------------
# This segment confirms that data exists and gives a first look at schema.

df = pd.read_csv("data/players_22.csv")

print(f"Shape: {df.shape}")
print("\nColumn sample:")
print(df.columns[:20].tolist())

df.head(3)
```

```python
# Segment 3: Define modeling features and validate missing values
# ---------------------------------------------------------------
# We focus on your original feature set and audit missingness before modeling.

features = ["overall", "potential", "wage_eur", "value_eur", "age"]
id_cols = ["sofifa_id", "short_name", "club_name", "player_positions"]
keep_cols = [c for c in id_cols + features if c in df.columns]

missing_report = df[features].isna().sum().to_frame("missing_count")
missing_report["missing_pct"] = (missing_report["missing_count"] / len(df) * 100).round(3)

print("Missing value report:")
display(missing_report)

# Duplicate check for entity ID.
if "sofifa_id" in df.columns:
    duplicate_ids = df["sofifa_id"].duplicated().sum()
    print(f"Duplicate sofifa_id rows: {duplicate_ids}")
```

```python
# Segment 4: Clean rows and verify numeric constraints
# ----------------------------------------------------
# We drop rows with missing modeling features and enforce numeric types.

clean = df[keep_cols].copy()
clean = clean.dropna(subset=features)

for col in features:
    clean[col] = pd.to_numeric(clean[col], errors="coerce")

clean = clean.dropna(subset=features).reset_index(drop=True)

print(f"Rows after cleaning: {len(clean):,}")
print(clean[features].describe().T)
```

```python
# Segment 5: Outlier analysis using IQR method
# --------------------------------------------
# This checks how strongly each feature is affected by extreme values.

def iqr_outlier_report(frame, cols):
    rows = []
    for col in cols:
        q1 = frame[col].quantile(0.25)
        q3 = frame[col].quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        n_out = ((frame[col] < lower) | (frame[col] > upper)).sum()
        rows.append({
            "feature": col,
            "q1": q1,
            "q3": q3,
            "iqr": iqr,
            "lower_fence": lower,
            "upper_fence": upper,
            "outlier_count": int(n_out),
            "outlier_pct": round(100 * n_out / len(frame), 2),
        })
    return pd.DataFrame(rows)

outlier_report = iqr_outlier_report(clean, features)
display(outlier_report)
```

```python
# Segment 6: Feature engineering + robust preprocessing
# -----------------------------------------------------
# Important for K-Means:
# - wage/value are heavy-tailed, so we apply log1p transform.
# - We then use RobustScaler to reduce outlier influence.

model_df = clean.copy()

# Log-transform monetary columns to compress extreme ranges.
for col in ["wage_eur", "value_eur"]:
    model_df[f"log_{col}"] = np.log1p(model_df[col])

model_features = ["overall", "potential", "age", "log_wage_eur", "log_value_eur"]
X_raw = model_df[model_features].copy()

scaler = RobustScaler()
X = scaler.fit_transform(X_raw)

print("Model features used:", model_features)
print("Scaled matrix shape:", X.shape)
```

```python
# Segment 7: Select K using multiple metrics
# ------------------------------------------
# We evaluate K from 2 to 10 with:
# - inertia (elbow heuristic)
# - silhouette score (higher is better)
# - Davies-Bouldin index (lower is better)

k_values = list(range(2, 11))
records = []

for k in k_values:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=50)
    labels = km.fit_predict(X)
    records.append({
        "k": k,
        "inertia": km.inertia_,
        "silhouette": silhouette_score(X, labels),
        "davies_bouldin": davies_bouldin_score(X, labels),
    })

metrics_df = pd.DataFrame(records)
display(metrics_df)

fig, axes = plt.subplots(1, 3, figsize=(18, 4))

sns.lineplot(data=metrics_df, x="k", y="inertia", marker="o", ax=axes[0])
axes[0].set_title("Elbow: Inertia vs K")

sns.lineplot(data=metrics_df, x="k", y="silhouette", marker="o", ax=axes[1])
axes[1].set_title("Silhouette vs K")

sns.lineplot(data=metrics_df, x="k", y="davies_bouldin", marker="o", ax=axes[2])
axes[2].set_title("Davies-Bouldin vs K")

plt.tight_layout()
plt.show()
```

```python
# Segment 8: Train final K-Means model
# ------------------------------------
# Pick K after reviewing plots above.
# You can change FINAL_K and rerun this segment.

FINAL_K = 5

final_km = KMeans(n_clusters=FINAL_K, random_state=RANDOM_STATE, n_init=100)
model_df["cluster"] = final_km.fit_predict(X)

print("Final silhouette:", round(silhouette_score(X, model_df["cluster"]), 4))
print("Final Davies-Bouldin:", round(davies_bouldin_score(X, model_df["cluster"]), 4))

model_df[["short_name", "club_name", "overall", "potential", "wage_eur", "value_eur", "age", "cluster"]].head(10)
```

```python
# Segment 9: Interpret clusters with aggregated statistics
# --------------------------------------------------------
# This is the most important portfolio part: explain cluster meaning.

cluster_profile = (
    model_df
    .groupby("cluster")[features]
    .agg(["count", "mean", "median", "min", "max"])
    .round(2)
)

display(cluster_profile)

# Show top players by value in each cluster to make clusters tangible.
for c in sorted(model_df["cluster"].unique()):
    print(f"\nCluster {c} - Top 5 by value_eur")
    display(
        model_df[model_df["cluster"] == c]
        .sort_values("value_eur", ascending=False)
        [["short_name", "club_name", "overall", "potential", "wage_eur", "value_eur", "age"]]
        .head(5)
    )
```

```python
# Segment 10: 2D visualization with PCA
# -------------------------------------
# PCA projection helps visually inspect separation quality between clusters.

pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X)

plot_df = model_df.copy()
plot_df["pc1"] = X_pca[:, 0]
plot_df["pc2"] = X_pca[:, 1]

plt.figure(figsize=(10, 7))
sns.scatterplot(data=plot_df, x="pc1", y="pc2", hue="cluster", palette="tab10", s=35, alpha=0.8)
plt.title("Player Clusters in PCA Space")
plt.legend(title="Cluster", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Segment 11: Save outputs for portfolio artifacts
# ------------------------------------------------
# Saving these files helps you showcase reproducible outputs in GitHub.

metrics_df.to_csv("reports/kmeans_k_selection_metrics.csv", index=False)
outlier_report.to_csv("reports/feature_outlier_report.csv", index=False)
model_df.to_csv("reports/players22_clustered.csv", index=False)

print("Saved:")
print("- reports/kmeans_k_selection_metrics.csv")
print("- reports/feature_outlier_report.csv")
print("- reports/players22_clustered.csv")
```
